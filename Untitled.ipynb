```python
import re
from IPython.display import HTML, display

def render_youtube_video(url: str, width: int = 560, height: int = 315):
    try:
        regex = r"(?:v=|\/)([0-9A-Za-z_-]{11})"
        match = re.search(regex, url)

        if not match:
            raise ValueError("Invalid YouTube URL")

        video_id = match.group(1)
        embed_url = f"https://www.youtube-nocookie.com/embed/{video_id}"

        iframe = f"""
        <iframe width="{width}" height="{height}"
        src="{embed_url}"
        title="YouTube video player"
        frameborder="0"
        allow="accelerometer; autoplay; clipboard-write; encrypted-media; gyroscope; picture-in-picture"
        allowfullscreen>
        </iframe>
        """

        display(HTML(iframe))
        return "success"

    except Exception as e:
        return e
```

```python
render_youtube_video("https://www.youtube.com/watch?v=G4A0nqoDvWA")
```

```python
import urllib.request
from IPython import display
```

```python
def is_valid(URL:str) ->bool:
    try:
        response_status=urllib.request.urlopen(URL).getcode()
        assert response_status==200
        return True
    except Exception as e:
        return False
```

```python
def render_site(URL:str, width:str="100%", height: str="600")->str:
    try:
        if is_valid(URL):
            response=display.IFrame(src=URL,width=width,height=height)
            display.display(response)
            return "success"
        else:
            raise "error"

    except Exception as e:
        raise e
```

```python
render_site("https://ragibcs.gitlab.io")
```
